## Alembic in FastAPI (Notes)

### 1. What is Alembic?

**Alembic** is a **database migration tool for SQLAlchemy**. In FastAPI projects, it is used to **manage and apply changes to the database schema over time**.

When developers modify database models (such as adding a column, changing a field type, or creating a new table), Alembic helps update the database structure without manually altering tables.

In simple terms, **Alembic works like version control for your database schema**, similar to how Git tracks changes in code.

---

### 2. Why Alembic is Important

During development, database models often change. Without a migration tool, developers would have to manually run SQL commands to modify the database.

Alembic solves this problem by:

* Tracking schema changes
* Generating migration scripts automatically
* Applying updates safely to the database
* Maintaining a history of schema versions

---

### 3. Basic Workflow of Alembic

Typical steps when using Alembic in a FastAPI project:

1. **Initialize Alembic**
2. **Create a migration script**
3. **Apply the migration to the database**

---

### 4. Example

#### Step 1: Install Alembic

```bash
pip install alembic
```

---

#### Step 2: Initialize Alembic

```bash
alembic init -t async migrations
```

This creates an **alembic folder** and configuration files for managing migrations.

---

#### Step 3: Example Model

```python
from sqlmodel import SQLModel, Field

class Seller(SQLModel, table=True):
    id: int = Field(primary_key=True)
    name: str
```

Now suppose we modify the model by adding a new column:

```python
class Seller(SQLModel, table=True):
    id: int = Field(primary_key=True)
    name: str
    email: str
```

---

#### Step 4: Generate Migration

```bash
alembic revision --autogenerate -m "add email column"
```

Alembic will create a migration file describing the schema change.

---

#### Step 5: Apply Migration

```bash
alembic upgrade head
```

This updates the database by adding the **email column** to the table.

---

### 5. What Alembic Can Do

Alembic helps with:

* Creating new tables
* Adding or removing columns
* Changing column types
* Creating indexes
* Managing foreign keys
* Tracking database schema history

---

### 6. Short Summary

Alembic is a **database migration tool used with SQLAlchemy in FastAPI projects**. It tracks database schema changes and allows developers to update the database structure safely using migration scripts instead of manual SQL commands.
